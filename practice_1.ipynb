````python
#### Опис данних
import os
import pathlib
import pandas as pd

DIR = os.getcwd()
DATA_PATH = pathlib.Path(DIR) / "parsed.csv"

if not DATA_PATH.exists():
    raise FileNotFoundError(f"CSV файл не знайдено по шляху: {DATA_PATH}")

df = pd.read_csv(DATA_PATH)
print("Зразок таблиці:", df.head(), sep="\n")
print("\n", "Статистичний опис таблиці:", df.describe(), sep="\n")
print("\n", "Імена колонок:", df.columns, sep="\n")
print("\n", "Розмір таблиці:", df.shape, sep="\n")
````

#### Вправа 1

Знайдіть 95-й процентиль магнітуди землетрусів у Японії, використовуючи тип магнітуди mb . Фільтруйте дані за ключовими словами в колонці place
(наприклад, містить "Japan") та типом магнітуди. Використовуйте метод quantile(0.95) для обчислення процентиля. Виведіть результат з коментарем.

````python
df_1 = df.copy()

# Видалення рядків з пропущеними значеннями у колонці "mag"
df_1 = df_1.dropna(subset=["mag"])

# Пошук тільки по Японії та по магнітуді типу "mb"
df_1 = df_1[df_1["parsed_place"].str.contains("japan", na=False, case=False) & (df_1["magType"] == "mb")]

mag_0_95 = df_1["mag"].quantile([0.95])
print("95 перцентиль магнітуди землетрусів у Японії (тип 'mb') складає:", str(mag_0_95.loc[0.95]))

# очистка пам'яті
del df_1
````

**4. Як додати новий стовпець на основі умовної логіки (наприклад, за допомогою np.where або apply )? Наведіть приклад для перевірки наявності
цунамі.**

**Відповідь:**

```python

import pandas as pd
import numpy as np

# Приклад датасету землетрусів (tsunami: 1 - було, 0 - не було)
df = pd.DataFrame({
    'magnitude': [5.1, 7.8, 6.2, 8.1],
    'tsunami': [0, 1, 0, 1]
})

# Спосіб 1: np.where (найшвидший і рекомендований)
df['has_tsunami'] = np.where(df['tsunami'] == 1, 'Так', 'Ні')

# Спосіб 2: apply з lambda-функцією (для складніших умов, наприклад, за магнітудою та прапорцем)
df['tsunami_risk'] = df.apply(
    lambda row: 'Високий ризик' if (row['tsunami'] == 1 and row['magnitude'] >= 7.5) else 'Низький/Немає',
    axis=1
)
```

**7. Яке значення має колонка tsunami (0/1 або True/False), і як це впливає на підрахунок кількості цунамі?**

**Відповідь:**

 Колонка `tsunami` має значення 0 або 1 (або еквівалентно False/True). Це означає, що для підрахунку кількості цунамі 
 можна просто сумувати значення цієї колонки: 1 відповідає наявності цунамі, 0 - його відсутності.

```python

# або
count_tsunami = (df['tsunami'] == 1).sum()
```
